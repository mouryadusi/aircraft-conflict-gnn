```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("default")

print("Ready.")
```

```python
comparison = pd.read_csv("../processed_data/model_comparison.csv")

comparison
```

```python
plt.figure(figsize=(7,5))

plt.bar(
    comparison["Model"],
    comparison["F1"]
)

plt.ylabel("F1 Score")
plt.title("Model Comparison (F1 Score)")

plt.tight_layout()

plt.savefig(
    "../results/model_f1_comparison.png",
    dpi=300
)

plt.show()

print("Saved model_f1_comparison.png")
```

```python
comparison_plot = comparison.copy()

comparison_plot = comparison_plot.dropna()

comparison_plot.set_index("Model")[["Precision","Recall"]].plot(
    kind="bar",
    figsize=(8,5)
)

plt.ylabel("Score")

plt.tight_layout()

plt.savefig(
    "../results/model_precision_recall.png",
    dpi=300
)

plt.show()

print("Saved model_precision_recall.png")
```

```python
roc = comparison.dropna()

plt.figure(figsize=(7,5))

plt.bar(
    roc["Model"],
    roc["ROC_AUC"]
)

plt.ylabel("ROC-AUC")
plt.title("ROC-AUC Comparison")

plt.tight_layout()

plt.savefig(
    "../results/model_roc_auc.png",
    dpi=300
)

plt.show()

print("Saved model_roc_auc.png")
```
