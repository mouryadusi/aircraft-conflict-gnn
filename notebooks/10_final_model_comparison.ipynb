```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

RESULTS = "/Users/Mourya/projects/aircraft-conflict-gnn/notebooks/results"

os.makedirs(RESULTS, exist_ok=True)
```

```python
comparison = pd.DataFrame({

    "Model":[
        "Rule-based",
        "XGBoost",
        "GCN",
        "GAT"
    ],

    "Precision":[
        0.0555,
        0.6119,
        0.0000,
        0.0048
    ],

    "Recall":[
        0.3402,
        0.8367,
        0.0000,
        0.3333
    ],

    "F1":[
        0.0955,
        0.7069,
        0.0000,
        0.0094
    ],

    "ROC_AUC":[
        np.nan,
        0.9998,
        0.5000,
        0.9420
    ]

})

comparison
```

```python
comparison.to_csv(

    os.path.join(
        RESULTS,
        "final_model_comparison.csv"
    ),

    index=False

)
```

```python
plt.figure(figsize=(7,5))

plt.bar(

    comparison["Model"],
    comparison["F1"]

)

plt.ylabel("F1 Score")

plt.title("Model Comparison (F1 Score)")

plt.grid(axis="y")

plt.savefig(

    os.path.join(
        RESULTS,
        "comparison_f1.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()
```

```python
x=np.arange(len(comparison))

width=0.35

plt.figure(figsize=(8,5))

plt.bar(

    x-width/2,

    comparison["Precision"],

    width,

    label="Precision"

)

plt.bar(

    x+width/2,

    comparison["Recall"],

    width,

    label="Recall"

)

plt.xticks(

    x,

    comparison["Model"]

)

plt.legend()

plt.grid(axis="y")

plt.title("Precision vs Recall")

plt.savefig(

    os.path.join(

        RESULTS,

        "precision_recall_comparison.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()
```

```python
roc = comparison.dropna()

plt.figure(figsize=(7,5))

plt.bar(

    roc["Model"],

    roc["ROC_AUC"]

)

plt.ylim(0,1)

plt.ylabel("ROC-AUC")

plt.grid(axis="y")

plt.title("ROC-AUC Comparison")

plt.savefig(

    os.path.join(

        RESULTS,

        "roc_auc_comparison.png"

    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()
```
